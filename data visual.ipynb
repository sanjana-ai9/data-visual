```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Month": ["January", "February", "March", "April", "May", "June"],
    "Sales": [12000, 15000, 18000, 14000, 22000, 25000],
    "Product": ["Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet"]
}

df = pd.DataFrame(data)

print(df)
```

```python
print("Sales Dataset:")
print(df)

print("\nDataset Information:")
print(df.info())

print("\nTotal Sales:", df["Sales"].sum())

print("Average Sales:", df["Sales"].mean())
```

```python
plt.figure(figsize=(8, 5))

plt.bar(df["Month"], df["Sales"])

plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales Amount")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(df["Month"], df["Sales"], marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales Amount")

plt.show()
```

```python
highest_sales = df.loc[df["Sales"].idxmax()]
lowest_sales = df.loc[df["Sales"].idxmin()]

print("Highest Sales:")
print(highest_sales)

print("\nLowest Sales:")
print(lowest_sales)

print("\nObservations:")

print("1. Sales increased from January to June overall.")

print("2. June recorded the highest sales.")

print("3. January recorded the lowest sales.")
```
